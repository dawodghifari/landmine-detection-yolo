```python
# '/Users/dghifari/02-University/SEM-1-2025/ELEC3612/Project 2/trial 2/yolo11-landmine/results.csv'

import pandas as pd
from bokeh.io import output_notebook, output_file, show
from bokeh.plotting import figure
from bokeh.layouts import gridplot

output_file("file.html")

df = pd.read_csv('/Users/dghifari/02-University/SEM-1-2025/ELEC3612/Project 2/trial 2/yolo11-landmine/results.csv' )

window = 10

train_metrics = [
    'train/box_loss',
    'train/cls_loss',
    'train/dfl_loss',
    'metrics/precision(B)',
    'metrics/recall(B)'
]
val_metrics = [
    'val/box_loss',
    'val/cls_loss',
    'val/dfl_loss',
    'metrics/mAP50(B)',
    'metrics/mAP50-95(B)'
]

plots = []
for metric in train_metrics:
    p = figure(title=metric, x_axis_label='epoch', y_axis_label=metric, width=300, height=300)
    p.line(df['epoch'], df[metric], legend_label='raw')
    p.line(df['epoch'], df[metric].rolling(window, min_periods=1).mean(), line_dash='dotted', legend_label='smooth')
    plots.append(p)

for metric in val_metrics:
    p = figure(title=metric, x_axis_label='epoch', y_axis_label=metric, width=300, height=300)
    p.line(df['epoch'], df[metric], legend_label='raw')
    p.line(df['epoch'], df[metric].rolling(window, min_periods=1).mean(), line_dash='dotted', legend_label='smooth')
    plots.append(p)

grid = gridplot([plots[:5], plots[5:]], toolbar_location='right')

show(grid)


```
